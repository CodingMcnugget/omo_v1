```python
# -*- coding: utf-8 -*-
import cv2
from ultralytics import YOLO

def main():
    # 1. 加载官方 COCO 预训练模型（包含 "bottle" 类别）
    model = YOLO("yolo12n.pt")  # 确保你本地已经 pip 安装了 ultralytics 并下载好了 yolov8n.pt
    
    # 2. 打开系统摄像头
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("无法打开摄像头")
        return
    
    # （可选）调低分辨率以获得更高的推理速度
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)
    
    # 3. 获取类别名称表。COCO 模型里，索引 39 对应 "bottle"
    class_names = model.names  # 例如 {0:"person", 1:"bicycle", ..., 39:"bottle", ...}

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # 4. 推理
        results = model(frame)      # results 是一个列表，我们只需要取 results[0]
        dets = results[0]
        
        found_bottle = False
        # 5. 遍历所有检测框，看看有没有 cls_name == "bottle"
        for box in dets.boxes:
            cls_id = int(box.cls[0].cpu().numpy())
            cls_name = class_names[cls_id]
            if cls_name == "bottle":
                found_bottle = True
                # 绘制水瓶的检测框（可视化用）
                x1, y1, x2, y2 = box.xyxy[0].cpu().numpy().astype(int)
                conf = box.conf[0].cpu().numpy()
                label = f"bottle {conf:.2f}"
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
                ((tw, th), _) = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 1)
                cv2.rectangle(frame, (x1, y1 - th - 8), (x1 + tw, y1), (0, 255, 0), -1)
                cv2.putText(frame, label, (x1, y1 - 5),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1, cv2.LINE_AA)

        # 6. 如果检测到水瓶，就在画面上或控制台提示
        if found_bottle:
            cv2.putText(frame, "Detected: Water Bottle", (20, 40),
                        cv2.FONT_HERSHEY_SIMPLEX, 1.0, (0, 0, 255), 2, cv2.LINE_AA)

        # 7. 显示结果
        cv2.imshow("Water Bottle Detection", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
